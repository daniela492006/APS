### a) Circuito a — Filtro paso bajo

**1. Función de transferencia.** La bobina y la resistencia están en serie; la salida se mide sobre el condensador. Por divisor de tensión:

$$H_a(s)=\frac{V_2(s)}{V_1(s)}=\frac{Z_C}{sL+R+Z_C}=\frac{1}{LCs^2+RCs+1}.$$

Dividiendo por $LC$ se identifica el denominador estándar $s^2+(\omega_0/Q)s+\omega_0^2$, con $\omega_0^2=1/(LC)$ y $\omega_0/Q=R/L$. Así:

$$L=\frac{RQ}{\omega_0}=\frac1{\sqrt2}\ \mathrm H,\qquad C=\frac1{\omega_0^2L}=\sqrt2\ \mathrm F.$$

$$\boxed{H_a(s)=\frac1{s^2+\sqrt2s+1}}$$

**2. Respuesta de módulo y fase.** Sustituyendo $s=j\omega$:

$$H_a(j\omega)=\frac1{1-\omega^2+j\sqrt2\omega},\qquad |H_a(j\omega)|=\frac1{\sqrt{(1-\omega^2)^2+2\omega^2}}.$$

$$\phi_a(\omega)=-\operatorname{atan2}(\sqrt2\omega,1-\omega^2).$$

El módulo en decibelios es $M_a(\omega)=20\log_{10}|H_a(j\omega)|$. La fase se representa en grados multiplicando la expresión anterior por $180/\pi$. La aproximación asintótica del módulo es $0$ dB para $\omega<1$ y $-40\log_{10}\omega$ dB para $\omega>1$.

![Gráfica de Bode del circuito a](graficas/a_paso_bajo.png)

**3. Tipo de filtro y ancho de banda.** Es un **paso bajo de segundo orden**: $|H_a(0)|=1$ y $|H_a(j\omega)|\to0$ para $\omega\to\infty$. En $\omega_c=1\ \mathrm{rad/s}$ el módulo es $1/\sqrt2$ ($-3.01$ dB). La banda de paso de $-3$ dB es $0\le\omega\le1$ y su **ancho de banda es $BW_a=1\ \mathrm{rad/s}$**.


### b) Circuito b — Filtro paso alto con realce

**1. Función de transferencia.** La resistencia y el condensador están en paralelo entre la entrada y la salida; la bobina está conectada a tierra en la salida. La impedancia entre entrada y salida es $Z_{RC}=R/(1+sRC)$, por lo que:

$$H_b(s)=\frac{sL}{sL+R/(1+sRC)}=\frac{LCs^2+(L/R)s}{LCs^2+(L/R)s+1}.$$

Al dividir por $LC$:

$$H_b(s)=\frac{s^2+s/(RC)}{s^2+s/(RC)+1/(LC)}.$$

Se cumple $\omega_0^2=1/(LC)$ y $\omega_0/Q=1/(RC)$. Por tanto:

$$C=\frac{Q}{R\omega_0}=\frac1{\sqrt2}\ \mathrm F,\qquad L=\frac1{\omega_0^2 C}=\sqrt2\ \mathrm H.$$

$$\boxed{H_b(s)=\frac{s^2+\sqrt2s}{s^2+\sqrt2s+1}}$$

**2. Respuesta de módulo y fase.** Al sustituir $s=j\omega$:

$$H_b(j\omega)=\frac{-\omega^2+j\sqrt2\omega}{1-\omega^2+j\sqrt2\omega},\qquad |H_b(j\omega)|=\sqrt{\frac{\omega^4+2\omega^2}{(1-\omega^2)^2+2\omega^2}}.$$

$$\phi_b(\omega)=\operatorname{atan2}(\sqrt2\omega,-\omega^2)-\operatorname{atan2}(\sqrt2\omega,1-\omega^2).$$

El módulo en dB es $M_b=20\log_{10}|H_b|$. A bajas frecuencias, $|H_b|\approx\sqrt2\omega$ ($+20$ dB/década); a altas frecuencias, $|H_b|\to1$ ($0$ dB). El cero adicional en $s=-\sqrt2$ se considera en la aproximación asintótica.

![Gráfica de Bode del circuito b](graficas/b_paso_alto.png)

**3. Tipo de filtro y ancho de banda.** Es un **paso alto con realce**, porque atenúa las bajas frecuencias, presenta un máximo superior a $0$ dB y tiende a $0$ dB a altas frecuencias. Tomando como referencia la ganancia asintótica unitaria, la frecuencia de corte inferior satisface $|H_b(j\omega_c)|^2=1/2$:

$$\frac{\omega_c^4+2\omega_c^2}{1+\omega_c^4}=\frac12\ \Longrightarrow\ \omega_c^4+4\omega_c^2-1=0.$$

$$\boxed{\omega_c=\sqrt{\sqrt5-2}\approx0.48587\ \mathrm{rad/s}}$$

Como no existe corte superior finito, el **ancho de banda ideal es infinito**. Este corte se define respecto de la ganancia de alta frecuencia; el realce produce una región con ganancia superior a la unidad.


### c) Circuito c — Filtro paso banda

**1. Función de transferencia.** La resistencia está en serie con la entrada, y la salida se mide sobre la rama formada por $L$ y $C$ en paralelo:

$$Z_{LC}=\left(\frac1{sL}+sC\right)^{-1}=\frac{sL}{1+s^2LC}.$$

$$H_c(s)=\frac{Z_{LC}}{R+Z_{LC}}=\frac{sL}{R(1+s^2LC)+sL}=\frac{s/(RC)}{s^2+s/(RC)+1/(LC)}.$$

Como $\omega_0^2=1/(LC)$ y $\omega_0/Q=1/(RC)$:

$$C=\frac{Q}{R\omega_0}=\frac1{\sqrt2}\ \mathrm F,\qquad L=\frac1{\omega_0^2C}=\sqrt2\ \mathrm H.$$

$$\boxed{H_c(s)=\frac{\sqrt2s}{s^2+\sqrt2s+1}}$$

**2. Respuesta de módulo y fase.** Sustituyendo $s=j\omega$:

$$H_c(j\omega)=\frac{j\sqrt2\omega}{1-\omega^2+j\sqrt2\omega},\qquad |H_c(j\omega)|=\frac{\sqrt2\omega}{\sqrt{(1-\omega^2)^2+2\omega^2}}.$$

$$\phi_c(\omega)=\frac\pi2-\operatorname{atan2}(\sqrt2\omega,1-\omega^2).$$

El módulo en dB es $M_c=20\log_{10}|H_c|$. Las pendientes asintóticas son $+20$ dB/década para frecuencias bajas y $-20$ dB/década para frecuencias altas.

![Gráfica de Bode del circuito c](graficas/c_paso_banda.png)

**3. Tipo de filtro y ancho de banda.** Es un **paso banda**: la rama $LC$ en paralelo presenta impedancia máxima en resonancia, $\omega_0=1\ \mathrm{rad/s}$, donde $|H_c|=1$ ($0$ dB). Los cortes a $-3$ dB se obtienen imponiendo $|H_c|^2=1/2$:

$$\omega_1=\sqrt{\frac32}-\frac1{\sqrt2}\approx0.51764\ \mathrm{rad/s},\qquad \omega_2=\sqrt{\frac32}+\frac1{\sqrt2}\approx1.93185\ \mathrm{rad/s}.$$

$$\boxed{BW_c=\omega_2-\omega_1=\sqrt2\approx1.41421\ \mathrm{rad/s}}$$


## 3. Análisis de resultados
Los resultados muestran que cambiar la disposición de $R$, $L$ y $C$ modifica el comportamiento del circuito, incluso conservando $\omega_0=1\ \mathrm{rad/s}$ y $Q=1/\sqrt2$.

El **circuito a** presenta ganancia próxima a $0$ dB en bajas frecuencias y una caída de $-40$ dB/década en altas frecuencias, correspondiente a dos polos. En $\omega=1\ \mathrm{rad/s}$ alcanza $-3.01$ dB, por lo que se comporta como un paso bajo de segundo orden.

El **circuito b** atenúa las frecuencias bajas y tiende a $0$ dB en altas frecuencias, pero presenta un realce por encima de $0$ dB en la región de transición. Esto se debe al cero adicional del numerador, por lo que no debe confundirse con un paso alto Butterworth convencional. Su corte inferior, referido a la ganancia de alta frecuencia, es aproximadamente $0.48587\ \mathrm{rad/s}$.

El **circuito c** transmite principalmente alrededor de $\omega_0=1\ \mathrm{rad/s}$, con ganancia máxima de $0$ dB. Los cortes de $-3$ dB se sitúan en $0.51764$ y $1.93185\ \mathrm{rad/s}$, dando un ancho de banda de $\sqrt2\ \mathrm{rad/s}$.

Las curvas de fase concuerdan con la contribución de polos y ceros. Las aproximaciones asintóticas permiten interpretar las pendientes lejos de las frecuencias características, pero difieren de las curvas exactas cerca de las transiciones.

# 5. Bonus: Adaptación del circuito a 10 kHz

Se desea modificar los valores de los componentes del circuito propuesto para que opere a una frecuencia de $f_0=10\ \mathrm{kHz}$, manteniendo una resistencia de $R=50\ \Omega$.

La frecuencia angular correspondiente es:

$$
\omega_0=2\pi f_0=62831.85\ \mathrm{rad/s}
$$

### Cálculo de los componentes

Para el circuito adicional, las admitancias entre la entrada y la salida, y entre la salida y tierra, son:

$$
Y_s(s)=\frac{b}{sL}+saC
$$

$$
Y_p(s)=sC(1-a)+\frac{1-b}{sL}+\frac{1}{R}
$$

Aplicando el divisor de tensión, la función de transferencia es:

$$
H(s)=\frac{Y_s(s)}{Y_s(s)+Y_p(s)}
$$

Por tanto:

$$
\boxed{H(s)=\frac{b+aLCs^2}{1+\frac{L}{R}s+LCs^2}}
$$

La frecuencia de rechazo se obtiene anulando el numerador:

$$
b-aLC\omega_r^2=0
$$

De donde:

$$
\omega_r=\sqrt{\frac{b}{aLC}}
$$

Como el enunciado no especifica los parámetros $a$ y $b$, se adopta como elección de diseño:

$$
a=b=0.5
$$

De esta forma:

$$
\omega_r=\frac{1}{\sqrt{LC}}
$$

Para situar la frecuencia de rechazo en $f_0=10\ \mathrm{kHz}$, se debe cumplir:

$$
LC=\frac{1}{(2\pi f_0)^2}
$$

Se selecciona una capacitancia total de:

$$
C=225.079\ \mathrm{nF}
$$

Y se calcula la inductancia:

$$
L=\frac{1}{(2\pi f_0)^2C}
\approx 1.125395\ \mathrm{mH}
$$

Los valores de los componentes físicos se obtienen utilizando $a=b=0.5$.

Para las inductancias:

$$
L_1=\frac{L}{b}=2.25079\ \mathrm{mH}
$$

$$
L_2=\frac{L}{1-b}=2.25079\ \mathrm{mH}
$$

Para los condensadores:

$$
C_1=Ca=112.540\ \mathrm{nF}
$$

$$
C_2=C(1-a)=112.540\ \mathrm{nF}
$$

### Valores finales de los componentes
